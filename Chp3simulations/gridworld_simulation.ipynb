```python
import numpy as np
import matplotlib
matplotlib.use('Agg') 
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
```

```python
W = 5
H = 5
REWARDS = np.zeros((H,W))
REWARDS[0,1] = 10
REWARDS[0,3] = 5

V = np.zeros((H,W))
gamma = 0.99
actions = ["L","R","U","D"]
pi = [0.25,0.25,0.25,0.25]
```

```python
#Returns probability distribution(dict) of possible states and the reward
#state is a len 2 array (row,col)
#action is one of "L","R","U","D"
def get_transition(state, action):
    if state == (0,1):
        return {(4,1): 1.0}, 10
    elif state == (0,3):
        return {(2,3): 1.0}, 5
    elif (state[0] == 0 and action == "U") or (state[0] == H-1 and action == "D") or (state[1] == 0 and action == "L") or (state[1] == W-1 and action == "R"):
        return {state: 1.0}, -1
    else: 
        if state[0] == H-1:
            if action == "R":
                return {(state[0],state[1]+1): 1.0}, 0
            elif action == "L":
                return {(state[0],state[1]-1): 1.0}, 0
            elif action == "U":
                return {(state[0]-1,state[1]): 1.0}, 0
        else:
            if action == "R":
                return {(state[0],state[1]+1): 1.0}, 0
            elif action == "L":
                return {(state[0],state[1]-1): 1.0}, 0
            elif action == "U":
                return {(state[0]-1,state[1]): 1.0}, 0
            elif action == "D":
                return {(state[0]+1,state[1]): 1.0}, 0

    
    
```

```python
def calc_v(state):
    new_v = 0
    for i in range(len(actions)):
        probability_dict, reward = get_transition(state, actions[i])
        for next_state in probability_dict:
            new_v += pi[i] * probability_dict[next_state] * (reward + gamma * V[next_state[0]][next_state[1]])
    return new_v
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
cax = ax.imshow(V, cmap='inferno', vmin=-3, vmax=10) 
plt.colorbar(cax)
title = ax.set_title("Iteration 0")

# 1. THE DUMMY INITIALIZER (Stops Matplotlib's ghost frame)
def init():
    cax.set_array(np.zeros((H, W)))
    title.set_text("Iteration 0")
    return cax, title

# 2. THE ACTUAL MATH
def update(frame):
    global V 
    new_V = np.copy(V)
    
    for row in range(H):
        for col in range(W):
            new_V[row, col] = calc_v((row, col))
            
    V = new_V
    cax.set_array(V) 
    title.set_text(f"Iteration {frame + 1}")
    return cax, title


# 3. RENDER (Notice the init_func parameter)
print("Rendering...")
ani = FuncAnimation(fig, update, frames=20, init_func=init, blit=True)
ani.save("gridworld_learning.gif", writer=PillowWriter(fps=5))
print("Done")
```
